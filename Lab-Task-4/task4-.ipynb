```python
def _luhn_sum(s):
    digits = [int(ch) for ch in s if ch.isdigit()]
    digits.reverse()
    total = 0
    for i, d in enumerate(digits):
        if i % 2 == 1:
            d = d * 2
            if d > 9:
                d -= 9
        total += d
    return total

def is_valid_luhn(s):
    return _luhn_sum(s) % 10 == 0

def compute_check_digit(partial):
    total = _luhn_sum(partial + "0")
    return (10 - (total % 10)) % 10

def main():
    import sys
    if len(sys.argv) > 1:
        for arg in sys.argv[1:]:
            print(arg, "=>", "VALID" if is_valid_luhn(arg) else "INVALID")
    else:
        partial = "7992739871"
        cd = compute_check_digit(partial)
        full = partial + str(cd)
        print("Example full number:", full)
        print("Is valid:", is_valid_luhn(full))
        s = input("Enter a number to validate (or press Enter to skip): ").strip()
        if s:
            print("VALID" if is_valid_luhn(s) else "INVALID")

if __name__ == "__main__":
    main()
```
