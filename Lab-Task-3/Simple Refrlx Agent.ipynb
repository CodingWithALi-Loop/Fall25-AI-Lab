###  Lab 3: Model-Based Reflex Agent (Beginner friendly). This program reads a few room temperatures and decides whether to turn the AC ON or OFF. It remembers the last action so it does not repeat the same command unnecessarily.


```python


class Agent:
    def __init__(self, target_temp=20):
        self.target_temp = target_temp
        self.current_temp = None
        self.previous_action = None
    def sense(self, temp):
        self.current_temp = temp
    def decide(self):
        if self.current_temp > self.target_temp:
            action = "ON"
        else:
            action = "OFF"
        if action == self.previous_action:
            return f"No change ({action})"
        else:
            self.previous_action = action
            return action + " AC"
    def act(self, room_name):
        result = self.decide()
        print(f"{room_name}: {self.current_temp}°C  ->  {result}")
def main():
    rooms = {
        "Living Room": 20,
        "Drawing Room": 22,
        "Kitchen": 34,
        "Store Room": 26,
        "Bed Room": 24,
    }
    agent = Agent(target_temp=16)
    for name, temp in rooms.items():
        agent.sense(temp)
        agent.act(name)
    print("\nTarget temperature:", agent.target_temp)


if __name__ == "__main__":
    main()
```
