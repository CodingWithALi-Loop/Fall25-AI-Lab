```python
def extract_number(series):
    return (
        series.astype(str)
              .str.extract(r'(\d+)')   # extract digits
              .fillna(0)
              .astype(int)
    )
```

```python
def convertFloatToInt(df, columns):
    for col in columns:
        df[col] = extract_number(df[col])
```

```python
from sklearn.preprocessing import LabelEncoder

encoders = {}

def convertObjToInt(df, columns):
    for col in columns:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))
        encoders[col] = le   # store encoder for future use
```

```python
numeric_string_cols = [
    'ram_gb','ssd','hdd','graphic_card_gb',
    'warranty','processor_gnrtn','os_bit'
]

convertFloatToInt(df, numeric_string_cols)
```

```python
categorical_cols = [
    'brand','processor_brand','processor_name','ram_type',
    'os','weight','Touchscreen','msoffice','rating'
]

convertObjToInt(df, categorical_cols)
```
