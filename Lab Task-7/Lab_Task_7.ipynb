```python
from collections import deque
class Graph:
    def __init__(self, adjacency_list):
        self.adjacency_list = adjacency_list
    def get_neighbors(self, v):
        return self.adjacency_list[v]
    def h(self, n):
        H = {
            'A': 1,
            'B': 1,
            'C': 1,
            'D': 0  
        }
        return H[n]
    def a_star_algorithm(self, start_node, stop_node):
        open_list = set([start_node])
        closed_list = set([])
        g = {}
        g[start_node] = 0
        parents = {}
        parents[start_node] = start_node
        print(f"Start A*: start={start_node}, goal={stop_node}")
        while len(open_list) > 0:
            n = None
            for v in open_list:
                if n is None or g[v] + self.h(v) < g[n] + self.h(n):
                    n = v
            if n is None:
                print('Path does not exist!')
                return None
            print(f"Selected node: {n} with f={g[n] + self.h(n)} (g={g[n]}, h={self.h(n)})")
            if n == stop_node:
                reconst_path = []
                while parents[n] != n:
                    reconst_path.append(n)
                    n = parents[n]
                reconst_path.append(start_node)
                reconst_path.reverse()
                print('Path found: {}'.format(reconst_path))
                return reconst_path
            for (m, weight) in self.get_neighbors(n):
                print(f"  Checking neighbor {m} (weight={weight})")
                if m not in open_list and m not in closed_list:
                    open_list.add(m)
                    parents[m] = n
                    g[m] = g[n] + weight
                    print(f"    Adding {m} to open_list with g={g[m]} parent={parents[m]}")
                else:
                    if g[m] > g[n] + weight:
                        g[m] = g[n] + weight
                        parents[m] = n
                        print(f"    Found better path to {m}: g updated to {g[m]} parent={parents[m]}")
                        if m in closed_list:
                            closed_list.remove(m)
                            open_list.add(m)
                            print(f"    Moved {m} from closed_list back to open_list")
            open_list.remove(n)
            closed_list.add(n)
            print(f"  Moved {n} from open_list to closed_list")
        print('Path does not exist!')
        return None
adjacency_list = {
    'A': [('B', 90), ('C', 1), ('D', 17)],
    'B': [('D', 6)],
    'C': [('D', 2)],
    'D': []
}
graph1 = Graph(adjacency_list)
graph1.a_star_algorithm('A', 'C')
```
