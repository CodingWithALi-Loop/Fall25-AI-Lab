### `1. DFS with Stack or with stack`

```python
def dfs_stack(graph, start):
    visited = set()
    stack = [start]
    while stack:
        x = stack.pop()
        if x not in visited:
            print(x, end=" ")
            visited.add(x)
            for neighbor in reversed(graph.get(x, [])):
                if neighbor not in visited:
                    stack.append(neighbor)

def dfs_recursive(graph, node, visited=None):
    if visited is None:
        visited = set()
    if node in visited:
        return
    print(node, end=" ")
    visited.add(node)
    for neighbor in graph.get(node, []):
        if neighbor not in visited:
            dfs_recursive(graph, neighbor, visited)

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

dfs_stack(graph, 'A')
print()
dfs_recursive(graph, 'A')
print()
```
